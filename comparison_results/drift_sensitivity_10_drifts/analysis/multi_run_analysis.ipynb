```python
import pandas as pd


df = pd.read_csv('all_runs_combined.csv')
```

```python
with open("summary_tables.md", "w") as f:
    for dataset_name in df['dataset'].unique():
        f.write(f"## Dataset: {dataset_name}\n\n")
        df_subset = df[df['dataset'] == dataset_name]
        summary = df_subset.groupby(['batch_size', 'technique']).agg({
            'accuracy': ['mean', 'std'],
            'f1': ['mean', 'std'],
            'precision': ['mean', 'std'],
            'recall': ['mean', 'std'],
            'num_drifts' : ['mean'],
            'num_batches' : ['mean'],
        }).reset_index().sort_values(('f1', 'mean'), ascending=False)
        f.write(summary.to_markdown(index=False))
        f.write("\n\n---\n\n")
```

```python
# Write a summary table for all data combined (not split by dataset)
with open("summary_tables_all.md", "w") as f:
    summary = df.groupby(['batch_size', 'technique', 'dataset']).agg({
        'accuracy': ['mean', 'std'],
        'f1': ['mean', 'std'],
        'precision': ['mean', 'std'],
        'recall': ['mean', 'std'],
        'num_drifts' : ['mean'],
        'num_batches' : ['mean'],
    }).reset_index().sort_values(('f1', 'mean'), ascending=False)
    f.write("# Summary Table: All Datasets Combined\n\n")
    f.write(summary.to_markdown(index=False))
    f.write("\n\n---\n\n")
```

```python
# Boxplots of F1 scores for all techniques, split by dataset, saved in the parent folder
import matplotlib.pyplot as plt
import seaborn as sns
import os
summary = df.groupby(['batch_size', 'technique', 'dataset']).agg({
    'accuracy': ['mean', 'std'],
    'f1': ['mean', 'std'],
    'precision': ['mean', 'std'],
    'recall': ['mean', 'std'],
    'num_drifts' : ['mean'],
    'num_batches' : ['mean'],
}).reset_index()
parent_folder = os.path.abspath(os.path.join(os.getcwd(), os.pardir))

for dataset_name in summary['dataset'].unique():
    plt.figure(figsize=(12, 6))
    subset = summary[summary['dataset'] == dataset_name]
    sns.boxplot(x='technique', y=('f1', 'mean'), data=subset)
    plt.xticks(rotation=45)
    plt.title(f'Boxplot of Mean F1 Scores by Technique - {dataset_name}')
    plt.tight_layout()
    plot_path = os.path.join(parent_folder, f'f1_boxplot_{dataset_name}.png')
    plt.savefig(plot_path)
    plt.close()
    print(f"Boxplot for {dataset_name} saved to: {plot_path}")
```
